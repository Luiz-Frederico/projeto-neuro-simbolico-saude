## MLP Treinamento (Scikit-Learn + Tensorflow / Keras)


```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.callbacks import ModelCheckpoint
from tensorflow.keras.regularizers import l2
from sklearn.model_selection import train_test_split

# ============================================================
# 0. CARREGAMENTO, NORMALIZAÇÃO E DIVISÃO (ESTRATIFICADA)
# ============================================================
X = np.load("X_data.npy")
y = np.load("y_labels.npy")

X_norm = X / np.array([43.0, 180.0])

X_train, X_val, y_train, y_val = train_test_split(
    X_norm, y,
    test_size=0.15,
    shuffle=True,
    random_state=42,
    stratify=y  # Mantém a proporção exata das novas classes no treino e validação
)

# ============================================================
# 1. ARQUITETURA DE ALTA CAPACIDADE COM REGULARIZAÇÃO L2
# ============================================================
model = tf.keras.Sequential([
    tf.keras.Input(shape=(2,)),

    tf.keras.layers.Dense(128, activation='relu', kernel_regularizer=l2(0.0001)),
    tf.keras.layers.Dense(64, activation='relu', kernel_regularizer=l2(0.0001)),
    tf.keras.layers.Dense(32, activation='relu', kernel_regularizer=l2(0.0001)),

    tf.keras.layers.Dense(6, activation='softmax')
])

# Mantendo a taxa de aprendizagem cirúrgica para costurar os penhascos de 60 e 100 BPM
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# ============================================================
# 2. CALLBACKS OTIMIZADOS
# ============================================================
checkpoint = ModelCheckpoint(
    'melhor_modelo.keras',
    monitor='val_accuracy',
    save_best_only=True,
    mode='max',
    verbose=1
)

# ============================================================
# 3. TREINAMENTO EXPANDIDO
# ============================================================
history = model.fit(
    X_train, y_train,
    epochs=1000,
    batch_size=64,
    validation_data=(X_val, y_val),
    callbacks=[checkpoint],
    verbose=1
)

# ============================================================
# 4. AVALIAÇÃO E SALVAMENTO DEFINITIVO
# ============================================================
model = tf.keras.models.load_model('melhor_modelo.keras')
loss, acc = model.evaluate(X_norm, y, verbose=0)
print(f"\n🎯 Nova Acurácia Global de Validação: {acc*100:.2f}%")

# Salvando o modelo
model.save("modelo_saude_6classes_l2_1000ep_v2.keras")
print("💾 'modelo_saude_6classes_l2_1000ep_v2.keras' salvo com sucesso!")
```

## Teste do Modelo

```python
# ============================================================
# 1. CARREGAMENTO DO MODELO E CONFIGURAÇÕES
# ============================================================

modelo_ia = tf.keras.models.load_model("modelo_saude_6classes_l2_1000ep_v2.keras")

mapeamento_classes = {
    0: "Hipotermia",
    1: "Saudável",
    2: "Febril",
    3: "Febre",
    4: "Febre Alta",
    5: "Risco Cardiovascular"
}

# ============================================================
# 2. PIPELINE HÍBRIDO NEURO-SIMBÓLICO AVANÇADO
# ============================================================
def classificar_paciente(temperatura, bpm):
    """
    Retorna: nome_da_classe, confianca_exibicao, array_probabilidades_reais
    """
    # -----------------------------------------------------------------------
    # PARTE A: A REDE NEURAL SEMPRE CALCULA OS BASTIDORES (PROBABILIDADE REAL)
    # ------------------------------------------------------------------------
    temp_norm = temperatura / 43.0
    bpm_norm = bpm / 180.0
    input_dados = np.array([[temp_norm, bpm_norm]])

    # Executa a inferência contínua do Keras nos bastidores
    probabilidades_reais = modelo_ia.predict(input_dados, verbose=0)[0]
    classe_id_ia = np.argmax(probabilidades_reais)

    # ------------------------------------------------------------------------
    # PARTE B: O FILTRO SIMBÓLICO
    # ------------------------------------------------------------------------

    # Regra Classe 0: Hipotermia Estrita
    if temperatura <= 35.00:
        return mapeamento_classes[0], 1.0, probabilidades_reais

    # Regra Classe 2: Febril Estrito
    if 37.50 <= temperatura <= 37.79:
        return mapeamento_classes[2], 1.0, probabilidades_reais

    # Regra Classe 3: Febre Estrita
    if 37.80 <= temperatura <= 38.99:
        return mapeamento_classes[3], 1.0, probabilidades_reais

    # Regra Classe 4: Febre Alta Estrita
    if temperatura >= 39.00:
        return mapeamento_classes[4], 1.0, probabilidades_reais

    # --- ZONA TÉRMICA ESTÁVEL (35.01°C a 37.49°C) ---
    # Regra Classe 1: Saudável Estrito
    if 60.0 <= bpm <= 100.0:
        return mapeamento_classes[1], 1.0, probabilidades_reais

    # Regra Classe 5: Risco Cardiovascular Estrito
    if bpm < 60.0 or bpm > 100.0:
        return mapeamento_classes[5], 1.0, probabilidades_reais

    # -----------------------------------------------------------------
    # PARTE C: FALLBACK DE SEGURANÇA
    # -----------------------------------------------------------------
    return mapeamento_classes[classe_id_ia], probabilidades_reais[classe_id_ia], probabilidades_reais


# ============================================================
# 3. BATERIA DE TESTES DAS FRONTEIRAS CRÍTICAS (AUTOMÁTICO)
# ============================================================
testes = [
    (35.00, 59),   # Limite Hipotermia com BPM Risco -> Deve dar Hipotermia (0)
    (35.00, 70),   # Limite Hipotermia com BPM Normal -> Deve dar Hipotermia (0)
    (35.01, 60),   # Início Saudável com BPM Normal -> Deve dar Saudável (1)
    (35.01, 59),   # Início Saudável com BPM Baixo -> Deve dar Risco Cardíaco (5)
    (36.00, 60),   # Estável Normal -> Saudável (1)
    (37.49, 100),  # Limite Superior Saudável -> Saudável (1)
    (37.49, 101),  # Limite Superior Saudável com BPM Alto -> Risco Cardíaco (5)
    (37.50, 60),   # Fronteira Térmica Superior -> Febril (2)
    (38.01, 89),   # Faixa de Febre -> Febre (3)
    (38.99, 105),  # Limite superior Febre -> Febre (3)
    (39.00, 75),   # Gatilho Febre Alta -> Febre Alta (4)
    (41.10, 100),  # Caso Crítico de Febre Alta -> Deve exibir (Hiperpirexia)
]

print("\n" + "="*75)
print("     EXECUÇÃO DOS TESTES NO SISTEMA HÍBRIDO (NEURO-SIMBÓLICO)")
print("="*75)

for temp, bpm in testes:
    classe, conf, todas_probs = classificar_paciente(temp, bpm)
    
    # Detalhe visual de enriquecimento para temperaturas acima de 41.0°C
    classe_exibicao = f"{classe} (Hiperpirexia)" if temp > 41.0 else classe
    
    print(f"🌡️ Entrada: {temp}°C | 💓 {bpm} BPM")
    print(f"🧠 DIAGNÓSTICO: {classe_exibicao}")
    print(f"🎯 Confiança/Garantia: {conf*100:.2f}%")
    print("-" * 75)


# ============================================================
# 4. LOOP DE TESTES MANUAIS INTERATIVOS (UI)
# ============================================================

print("\n" + "="*75)
print("\n--- Sistema de Classificação de Pacientes (6 níveis) ---")
print("  - Saudável: Temperaturas 35.1-37.4°C E BPM 60-100.")
print("  - Risco Cardiovascular: Temperatura normal, mas BPM fora da faixa.")
print("  - Demais classes: Baseadas exclusivamente na temperatura.")
print("  - Atenção: Temperatura de 35.0 com BPM fora da faixa é diagnosticada como ")
print("    Hipotermia, mas apresentam informações relevantes relacionadas ao Risco Cardiovascular.")
print("    Distribuição Completa é fundamental para uma análise precisa do diagnóstico.")
print("    Temperatura acima de 41.1 será diagnosticada como Febre Alta (Hiperpirexia).")
print("\nDigite os valores ou 'sair' para encerrar.")
print("="*75)

while True:
    val_temp = input("\nTemperatura (°C): ").strip().lower()
    if val_temp == 'sair':
        break

    try:
        # 1. Validação da Temperatura: Garante que tem no máximo 1 casa decimal
        val_temp = val_temp.replace(',', '.')  # Trata caso digitem com vírgula
        if '.' in val_temp and len(val_temp.split('.')[1]) > 1:
            print("⚠️ Erro: Insira a temperatura com no máximo uma casa decimal (ex: 35.1).")
            continue
        temp = float(val_temp)

        # 2. Validação do BPM: Garante que é um número inteiro absoluto
        val_fc = input("Frequência Cardíaca (BPM): ").strip()
        if not val_fc.isdigit():
            print("⚠️ Erro: A frequência cardíaca deve ser um número inteiro (ex: 99).")
            continue
        fc = float(val_fc)

        # Executa a classificação pelo pipeline neuro-simbólico
        classe, conf, todas_probs = classificar_paciente(temp, fc)

        # Aplicação cirúrgica do enriquecimento visual 
        classe_exibicao = f"{classe} (Hiperpirexia)" if temp > 41.0 else classe

        print(f"-----------------------")
        print(f"🧠 RESULTADO: {classe_exibicao}")
        print(f"🎯 Confiança da IA Neurosimbólica: {conf*100:.2f}%")
        print(f"📊 Distribuição Completa (Cálculo da IA nos Bastidores):")
        print(f"   Hipotermia: {todas_probs[0]*100:.1f}%")
        print(f"   Saudável: {todas_probs[1]*100:.1f}%")
        print(f"   Febril: {todas_probs[2]*100:.1f}%")
        print(f"   Febre: {todas_probs[3]*100:.1f}%")
        print(f"   Febre Alta: {todas_probs[4]*100:.1f}%")
        print(f"   Risco Cardiovascular: {todas_probs[5]*100:.1f}%")
        print(f"-----------------------")

    except ValueError:
        print("⚠️ Erro: Insira apenas números válidos.")
    except KeyboardInterrupt:
        break

print("\nSessão encerrada.")
```
